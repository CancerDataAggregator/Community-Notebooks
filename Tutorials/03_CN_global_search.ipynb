Two free-text terms like this means "must touch both" -- every subject here is connected to *both* kidney and vcf somewhere in their record, still searched across every column at once. 2,516 subjects, with 23,096 vcf files between them -- a much smaller, more specific number than my first search, and I got there without writing a single filter string.

Now I want to actually look at the `anatomic_site` column on these results, and the numbers don't initially add up: `anatomic_site = kidney` alone shows 48,264 -- more than ten times the 3,657 subjects I actually matched. That's a sign these breakdown counts aren't counting subjects at all.

They're not. `anatomic_site` (like `diagnosis` and `format`) is a column on the **file** table, not the `subject` table. When a file-table column gets added to a `summarize_subjects()` call, the breakdown is a count of **matching files** carrying each value -- across every file tied to every matching subject -- not a count of subjects. A relatively small number of subjects with many files each can easily outnumber the subject total many times over.

With that in mind, the breakdown is still informative: the single biggest value isn't "kidney" at all, it's a missing/null value (101,439 files) -- plenty of files simply don't have an anatomic_site recorded. After that, "kidney" itself is the largest real value (48,264 files), which makes sense. But then "blood" shows up at 20,398 files -- real data riding along on these same kidney-matched subjects. That's not a bug; a subject who matched on "kidney" through some connection (a diagnosis, a file description, whatever) can easily also have files like blood draws that have nothing to do with kidney anatomy specifically. I didn't ask for blood data, but now I know it's there -- I also now know these counts describe files, not people.

Further down this same list, "right kidney" and "left kidney" also show up, each more specific than the plain word I typed. That's a genuine anatomy match -- CDA's anatomic_site field is backed by a real medical ontology (UBERON), and searching a broad term like "kidney" automatically pulls in anything more specific that sits underneath it in that hierarchy. "Right kidney" and "left kidney" are two of those more-specific terms; there are dozens more sitting even deeper in that same branch that never happened to show up directly in my particular result set. I didn't need to know that hierarchy existed, let alone look anything up in it, for this to work.

So two different things are going on in this one breakdown: files tagging along on a subject for a reason unrelated to anatomy (blood), and files matching through real anatomical specificity (right/left kidney). If I want to know *why* a specific anatomic_site row matched -- the literal word, or something more specific rolling up -- I can ask CDA to show its work:

```python
summarize_subjects('kidney', 'vcf', add_columns=['anatomic_site', 'diagnosis', 'format'], add_extras=['all'])
```

Same subjects, same count -- `add_extras` doesn't change what matched, it just annotates *why*. A row with "right kidney" now shows that it matched because it rolls up to "kidney" in the ontology.

Worth knowing: this kind of automatic rollup currently only happens for `anatomic_site` and `disease` -- those are the two fields backed by a real external ontology (UBERON and ICD-O-3, respectively). Most other harmonized fields, like `sex` or `species`, are fully standardized too, but don't have a hierarchy to roll up through in the first place -- there's no such thing as a "more specific" sex. A few fields (`stage`, `treatment_type`, and others) aren't harmonized yet at all. So `add_extras` won't do anything interesting on those -- not because something's broken, but because there's nothing underneath them yet to show.
