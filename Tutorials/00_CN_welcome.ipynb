## About cdapython

With CDA you search by harmonized, common-language terms. Using simple function calls you can get information about the subjects, files, or specimens you care about, in a standard dataframe (or TSV) format you can open in Excel, feed into a pipeline, or upload to your favorite cloud resource.

`cdapython` has these core functions:

---
- **`tables()`** shows you what tables are available to search
- **`columns()`** shows you what columns are in each table, and what they mean
- **`column_values()`** returns a unique'd list of all values in a column, with counts
- **`summarize_subjects()`** returns a profile of subject data matching your search criteria
- **`summarize_files()`** returns a profile of file data matching your search criteria
- **`get_subject_data()`** returns full subject rows matching your search criteria
- **`get_file_data()`** returns full file rows matching your search criteria
- **`intersect_subject_results()`** / **`intersect_file_results()`** combine two result sets via intersection
- **`expand_subject_results()`** / **`expand_file_results()`** flatten a nested per-subject or per-file column into a plain table
---

All of these functions take several parameters to let you tune your results. Use the `help()` function inside a notebook to find out more about any of them, e.g. `help(columns)`.

## About CDA data

You can think of the CDA as a really, really enormous spreadsheet full of data. To search this spreadsheet, you select columns that have data you're interested in, then filter the rows down to only the values you care about.

<div class="cdanote" style="background-color:#b3e5d5;color:black;padding:20px;">

CDA currently pulls data from six sources:
<ul>
<li><b>The <a href="https://gdc.cancer.gov/">Genomic Data Commons</a> (GDC)</b></li>
<li><b>The <a href="https://proteomic.datacommons.cancer.gov/pdc/">Proteomic Data Commons</a> (PDC)</b></li>
<li><b>The <a href="https://imaging.datacommons.cancer.gov/">Imaging Data Commons</a> (IDC)</b></li>
<li><b>The <a href="https://dataservice.datacommons.cancer.gov/#/home">General Commons</a> (GC)</b></li>
<li><b>The <a href="https://caninecommons.cancer.gov/#/explore">Integrated Canine Data Commons</a> (ICDC)</b></li>
<li><b>The <a href="https://datacommons.cancer.gov/repository/clinical-and-translational-data-commons">Clinical and Translational Data Commons</a> (CTDC)</b></li>
</ul>

(Note: <a href="https://www.isb-cgc.org/">ISB-CGC</a> is a close partner of CDA -- it's the source of all mutation data in CDA, and separately consumes CDA's phenotypic data for its own tools -- but it isn't one of the six source data centers itself. See the ISB-CGC notebook in this set for more on that relationship.)

The CDA makes this data searchable in two main endpoints:

<ul>
<li><b>subject:</b> a patient entity capturing the study-independent metadata for research subjects. Human research subjects are usually not traceable to a particular person, to protect their privacy.</li>
<li><b>file:</b> a unit of data about subjects, research subjects, specimens, or their associated information.</li>
</ul>
</div>

If you're looking to build a cohort of distinct individuals who meet some criteria, you'd search using `get_subject_data()`, which returns a table with one row per subject -- then use the `add_columns` parameter to pull in extra information per subject (file details, treatment history, whatever's relevant to what you're building).
